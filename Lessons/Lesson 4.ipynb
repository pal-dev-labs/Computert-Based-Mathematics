# CONCETTO: STRUTTURE CONDIZIONALI
Un programma è un elenco di istruzioni che vengono eseguite in sequenza. 

Con le strutture condizionali, è possibile eseguire determinati blocchi di codice **CONDIZIONALMENTE** e/o **RIPETUTAMENTE**: questi blocchi di base possono essere combinati per creare programmi molto sofisticati.

## Riepilogo: Operazioni di confronto

Un tipo di operazione che può essere molto utile è il confronto tra valori diversi.
A tale scopo, Python implementa operatori di confronto standard, che restituiscono valori booleani ``True`` e ``False``.

Le operazioni di confronto sono elencate nella tabella seguente:

| ``a == b`` | ``a`` uguale a ``b``

| ``a != b`` | ``a`` diverso da ``b``

| ``a < b`` | ``a`` minore di ``b``

| ``a > b`` | ``a`` maggiore di ``b``

| ``a <= b`` | ``a`` minore o uguale a ``b``

|``a >= b`` | ``a`` maggiore o uguale a ``b``



## Conditional Statements: ``if``-``elif``-``else``:

Scriviamo una funzione Python che prende in input due liste e restituisce True se hanno almeno un elemento in comune.

```python
def checklist(lista1,lista2):
    found = False
    counter=0
    for i in lista1:
        for j in lista2:
            if i == j:
                counter=counter+1
                found = True
    return found,counter           
```

```python
x = [1,2,3,5,6]
y= [1,2,7,8,9,10]
a1, a2 = checklist(x,y)
print("Trovato stesso elemento:", a1, " N. volte =", a2)
```

# LA DISTANZA EUCLIDEA

1. Consideriamo due punto nel piano $A=(x1,y1)$ e $B=(x2,y2)$
2. La distanza euclidea tra i due punti $d(A,B)$ è data da $$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$
3. Un punto del piano $A(x,y)$ può essere associato ad un vettore $\vec v$ che ha come componenti le coordinate del punto $\vec v(x,y)$
4. I punti $A=(x1,y1)$ e $B=(x2,y2)$ possono quindi essere associati ai vettori $\vec v1(x_1,y_1), \vec v2(x_2,y_2)$
5. La differenza $\vec v1 - \vec v2$ è data da $ [(x_2 - x_1), (y_2 - y_1)]$
6. 


Iniziamo definendo due punti $A$ e $B$. 

In NumPy, li rappresentiamo come vettori posizione che partono dall'origine $(0,0)$:$\vec{a} = [x_1, y_1]$$\vec{b} = [x_2, y_2]$2. Il vettore differenza $(\vec{d} = \vec{b} - \vec{a})$Quando eseguiamo la sottrazione diff = B - A, stiamo creando un terzo vettore, $\vec{d}$. Geometricamente, questo vettore è la freccia che unisce la punta di $A$ alla punta di $B$.Questo vettore ha componenti:$$\vec{d} = [(x_2 - x_1), (y_2 - y_1)]$$È proprio questo vettore che rappresenta lo "scarto" o la "distanza direzionata" tra i due punti.3. Il Modulo Quadro (Norma al quadrato)Il modulo (o norma) di un vettore, indicato con $\|\vec{d}\|$, rappresenta la sua lunghezza. Il modulo quadro $\|\vec{d}\|^2$ è semplicemente il quadrato di quella lunghezza.In algebra vettoriale, il modulo quadro di un vettore è uguale al prodotto scalare del vettore per se stesso:$$\|\vec{d}\|^2 = \vec{d} \cdot \vec{d} = (x_2 - x_1)^2 + (y_2 - y_1)^2$$

```python
# punto_x1
x1 = np.array([1,2])   

# punto_x2
x2 = np.array([5,7])   
    
diff = x1-x2

print(diff)
```

```python
diff2 = diff**2
print(diff2)
```

```python
# 2. Calcolo della distanza Euclidea

# Eleviamo al quadrato le differenze, le sommiamo e facciamo la radice
distanza = np.sqrt(np.sum(diff2))
print(distanza)
```

# Alternativa: np.linalg.norm 

```python
distanza2 = np.linalg.norm(x1 - x2)
print(distanza2)
```

- Il corpo A si muove di moto parabolico. parte dal punto O(0m ,170m) con una velocita iniziale $\vec v_0=(12 m/s,7m/s)$. 
$$\vec x(t) = (12 t, y(t) = 170 + 7 t - \frac{1}{2} g t^2)$$
- Il corpo B si muove con la seguente legge oraria: $$\vec x(t) = (30 cos(1 t), y(t) = 25 sin(1 t))$$

## Deermina la distanza minima tra i due corpi nei prima 10 secondi


```python
from scipy.optimize import minimize_scalar

def distanza(t):
    # Coordinate Corpo A
    xa = 12 * t
    ya = 170 + 7 * t - 0.5 * 9.81 * t**2
    
    # Coordinate Corpo B
    xb = 30 * np.cos(1 * t)
    yb = 25 * np.sin(1 * t)
    
    # Ritorna la distanza euclidea
    return np.sqrt((xa - xb)**2 + (ya - yb)**2)

# Cerchiamo il minimo nell'intervallo [0, 10] secondi
risultato = minimize_scalar(distanza, bounds=(0, 10), method='bounded')

print(f"Distanza minima: {risultato.fun:.4f} metri")
print(f"Raggiunta al tempo t: {risultato.x:.4f} secondi")
```

```python
%matplotlib widget
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.widgets import Slider
from scipy.optimize import minimize_scalar

# 1. Funzioni orarie e parametri
g = 9.81
t_max = 10

def get_pos(t):
    # Corpo A
    xa = 12 * t
    ya = 170 + 7 * t - 0.5 * g * t**2
    # Corpo B
    xb = 30 * np.cos(1 * t)
    yb = 25 * np.sin(1 * t)
    return (xa, ya), (xb, yb)

def calcola_distanza(t):
    posA, posB = get_pos(t)
    return np.sqrt((posA[0] - posB[0])**2 + (posA[1] - posB[1])**2)

# --- Calcolo Minimo Globale ---
res = minimize_scalar(calcola_distanza, bounds=(0, t_max), method='bounded')
t_min, d_min = res.x, res.fun

# 2. Setup Grafico
fig, ax = plt.subplots(figsize=(9, 6))
plt.subplots_adjust(bottom=0.25) # Spazio per lo slider

# Pre-calcolo traiettorie per le scie
t_all = np.linspace(0, t_max, 500)
posA_all, posB_all = get_pos(t_all)
ax.plot(posA_all[0], posA_all[1], 'r-', alpha=0.2)
ax.plot(posB_all[0], posB_all[1], 'b-', alpha=0.2)

# Elementi mobili
pointA, = ax.plot([], [], 'ro', markersize=8, label='Corpo A')
pointB, = ax.plot([], [], 'bo', markersize=8, label='Corpo B')
line_dist, = ax.plot([], [], 'g--', alpha=0.6, label='Distanza')

# Annotazioni
txt = ax.text(0.05, 0.9, '', transform=ax.transAxes, bbox=dict(facecolor='white', alpha=0.7))
ax.axhline(0, color='black', lw=1, ls='-') # Terreno

ax.set_xlim(-40, 130)
ax.set_ylim(-200, 200)
ax.legend(loc='upper right')
ax.grid(True, alpha=0.3)

# 3. Creazione dello Slider
ax_slider = plt.axes([0.2, 0.1, 0.6, 0.03])
slider_t = Slider(ax_slider, 'Tempo (s)', 0, t_max, valinit=0, valstep=0.05)

# 4. Funzione di aggiornamento
def update(val):
    t = slider_t.val
    (xa, ya), (xb, yb) = get_pos(t)
    d = calcola_distanza(t)
    
    pointA.set_data([xa], [ya])
    pointB.set_data([xb], [yb])
    line_dist.set_data([xa, xb], [ya, yb])
    
    info = f'Tempo: {t:.2f} s\nDistanza: {d:.2f} m'
    if abs(t - t_min) < 0.05:
        info += f'\n¡VICINO AL MINIMO!'
    txt.set_text(info)
    
    fig.canvas.draw_idle()

slider_t.on_changed(update)

# Inizializzazione
slider_t.set_val(0)
plt.show()

print(f"Punto di massima vicinanza: {d_min:.3f} m a t = {t_min:.3f} s")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# 1. MOTORE RADAR (La parte che gli studenti non toccano)
# ==========================================================
class SistemaRadar:
    def __init__(self, n_oggetti=12):
        self.robot = np.array([0, 0])
        self.oggetti = np.random.uniform(-10, 10, (n_oggetti, 2))
        self.colori = []

    def visualizza(self):
        plt.figure(figsize=(8, 8))
        plt.scatter(0, 0, color='blue', s=200, marker='s', label='Robot')
        
        # Disegno cerchi di riferimento
        for r in [3, 6, 9]:
            cerchio = plt.Circle((0, 0), r, color='gray', fill=False, linestyle=':', alpha=0.5)
            plt.gca().add_patch(cerchio)

        # Disegno oggetti classificati dagli studenti
        for i, punto in enumerate(self.oggetti):
            plt.scatter(punto[0], punto[1], color=self.colori[i], s=100)
            plt.text(punto[0]+0.3, punto[1]+0.3, f"Obj {i}", fontsize=9)

        plt.title("Display Radar - Output Funzione Studente")
        plt.xlim(-11, 11); plt.ylim(-11, 11); plt.grid(True, alpha=0.2)
        plt.show()

# ==========================================================
# 2. AREA DI LAVORO STUDENTE
# ==========================================================
def analizza_prossimita(punti_rilevati):
    """
    MODIFICA QUESTA FUNZIONE.
    Input: 'punti_rilevati' è un array NumPy con le coordinate (x, y) di diversi oggetti.
    Output: Deve restituire una lista di colori (stringhe) per ogni oggetto.
    """
    lista_colori = []
    posizione_robot = np.array([0, 0])

    # --- INIZIO LOGICA STUDENTE ---
    for i in range(len(punti_rilevati)):
        punto_attuale = punti_rilevati[i]
        
        # ESERCIZIO 1: Calcolare la distanza tra posizione_robot e punto_attuale
        # Suggerimento: usa np.sqrt(np.sum((vettore1 - vettore2)**2))
        distanza = np.sqrt(np.sum((punto_attuale - posizione_robot)**2))
        
        # ESERCIZIO 2: Usare i controlli condizionati per decidere il colore
        if distanza < 3.0:
            colore = 'red'       # Pericolo immediato
        elif distanza < 6.0:
            colore = 'orange'    # Attenzione
        else:
            colore = 'green'     # Distanza di sicurezza
            
        lista_colori.append(colore)
    # --- FINE LOGICA STUDENTE ---
    
    return lista_colori

# ==========================================================
# 3. ESECUZIONE DEL GIOCO
# ==========================================================
# Inizializzo il sistema
mio_radar = SistemaRadar(n_oggetti=15)

# Chiamo la funzione scritta dagli studenti
colori_decisi = analizza_prossimita(mio_radar.oggetti)

# Passo i risultati al motore grafico
mio_radar.colori = colori_decisi
mio_radar.visualizza()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# ==========================================================
# 1. MOTORE RADAR (Scatola nera - Non modificare)
# ==========================================================
class RadarSimulato:
    def __init__(self, n_oggetti=15):
        self.n = n_oggetti
        self.robot = np.array([0, 0])
        # Posizioni iniziali casuali (x, y) tra -10 e 10
        self.oggetti = np.random.uniform(-10, 10, (self.n, 2))
        # Velocità casuali: ogni oggetto si muove in modo diverso
        self.velocita = np.random.uniform(-0.4, 0.4, (self.n, 2))
        
        # Inizializzazione Grafica
        self.fig, self.ax = plt.subplots(figsize=(8, 8))
        self.punti_grafici = self.ax.scatter([], [], s=100, edgecolors='black')
        
    def setup_estetica(self):
        """Configura il look del radar"""
        self.ax.set_xlim(-11, 11)
        self.ax.set_ylim(-11, 11)
        self.ax.set_facecolor('#f0f0f0')
        
        # Disegniamo il Robot al centro
        self.ax.scatter(0, 0, color='blue', s=250, marker='s', label='Robot (Base)', zorder=5)
        
        # Disegniamo i cerchi delle zone di sicurezza
        for r, etichetta in zip([3, 6, 9], ['Pericolo', 'Attenzione', 'Sicuro']):
            cerchio = plt.Circle((0, 0), r, color='gray', fill=False, linestyle='--', alpha=0.4)
            self.ax.add_patch(cerchio)
            self.ax.text(0, r + 0.2, etichetta, color='gray', fontsize=8, ha='center')

        self.ax.legend(loc='upper right')
        self.ax.grid(True, linestyle=':', alpha=0.6)

    def aggiorna_frame(self, frame):
        """Funzione chiamata ogni 500ms dal sistema"""
        # 1. Muove gli oggetti nel tempo
        self.oggetti += self.velocita
        
        # 2. Rimbalzo: se un oggetto esce dai confini del radar, inverte la marcia
        fuori_limite = (self.oggetti > 10.5) | (self.oggetti < -10.5)
        self.velocita[fuori_limite] *= -1
        
        # 3. CHIAMATA ALLA FUNZIONE DEGLI STUDENTI
        # Passiamo l'array delle posizioni correnti
        colori_aggiornati = analizza_prossimita(self.oggetti)
        
        # 4. Applichiamo i cambiamenti alla grafica
        self.punti_grafici.set_offsets(self.oggetti)
        self.punti_grafici.set_color(colori_aggiornati)
        
        return self.punti_grafici,

# ==========================================================
# 2. AREA DI LAVORO STUDENTE (La logica da implementare)
# ==========================================================
def analizza_prossimita(punti_rilevati):
    """
    Input: 'punti_rilevati' è una matrice NumPy (N righe, 2 colonne).
    Output: Una lista di stringhe (colori) lunga N.
    """
    lista_colori = []
    posizione_robot = np.array([0, 0])

    # --- GLI STUDENTI LAVORANO QUI ---
    for i in range(len(punti_rilevati)):
        punto_attuale = punti_rilevati[i]
        
        # Calcoliamo la distanza euclidea usando la logica vettoriale di NumPy
        # Vettore differenza -> Elevamento al quadrato -> Somma -> Radice
        distanza = np.sqrt(np.sum((punto_attuale - posizione_robot)**2))
        
        # Controllo delle zone con strutture condizionate
        if distanza < 3.0:
            colore = 'red'      # Zona di collisione
        elif distanza < 6.0:
            colore = 'orange'   # Zona di allerta
        else:
            colore = 'green'    # Zona libera
            
        lista_colori.append(colore)
    # --- FINE LOGICA STUDENTE ---
    
    return lista_colori

# ==========================================================
# 3. ESECUZIONE
# ==========================================================
# Creiamo l'istanza del radar
radar = RadarSimulato(n_oggetti=18)
radar.setup_estetica()

# Creiamo l'animazione (intervallo in millisecondi)
animazione = FuncAnimation(
    radar.fig, 
    radar.aggiorna_frame, 
    interval=500, 
    blit=True, 
    cache_frame_data=False
)

plt.title("📡 SISTEMA RADAR ATTIVO - Monitoraggio Prossimità")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

# ==========================================================
# 1. MOTORE RADAR (Gestione Movimento e Grafica)
# ==========================================================
class RadarDinamico:
    def __init__(self, n_oggetti=8):
        self.n = n_oggetti
        # Posizioni (vettori posizione)
        self.posizioni = np.random.uniform(-9, 9, (self.n, 2))
        # Spostamento (Delta) costante per ogni frame
        self.delta = np.random.uniform(-0.4, 0.4, (self.n, 2))
        
        self.fig, self.ax = plt.subplots(figsize=(8, 8))
        self.punti_grafici = self.ax.scatter([], [], s=150, edgecolors='white', zorder=3)
        
        # Creiamo etichette di testo per mostrare le coordinate X in tempo reale
        self.testi = [self.ax.text(0, 0, '', color='white', fontsize=9) for _ in range(self.n)]
        
    def setup_radar(self):
        self.ax.set_xlim(-11, 11)
        self.ax.set_ylim(-11, 11)
        self.ax.set_facecolor('#001500') # Verde scurissimo "militare"
        
        # Disegno delle zone di soglia
        for r, c in zip([3, 6, 9], ['#ff4d4d', '#ffa500', '#2eb82e']):
            cerchio = plt.Circle((0, 0), r, color=c, fill=False, linestyle='--', alpha=0.3, linewidth=2)
            self.ax.add_patch(cerchio)
        
        self.ax.scatter(0, 0, color='white', s=100, marker='+', label='Stazione Base')
        self.ax.grid(True, color='green', alpha=0.1)

    def aggiorna(self, frame):
        # SPOSTAMENTO: Applichiamo il delta alle posizioni (Operazione Vettoriale)
        self.posizioni += self.delta
        
        # Rimbalzo sui bordi
        mask = (self.posizioni > 10.5) | (self.posizioni < -10.5)
        self.delta[mask] *= -1
        
        # Chiamata alla funzione degli studenti
        nuovi_colori = analizza_prossimita(self.posizioni)
        
        # Aggiornamento grafico dei punti
        self.punti_grafici.set_offsets(self.posizioni)
        self.punti_grafici.set_color(nuovi_colori)
        
        # Aggiornamento delle etichette (mostriamo x e y arrotondati)
        for i, txt in enumerate(self.testi):
            txt.set_position((self.posizioni[i, 0] + 0.3, self.posizioni[i, 1] + 0.3))
            txt.set_text(f"x:{self.posizioni[i,0]:.1f}")
            txt.set_color(nuovi_colori[i])

        return [self.punti_grafici] + self.testi

# ==========================================================
# 2. LOGICA STUDENTE (Ciclo for e Controlli)
# ==========================================================
def analizza_prossimita(punti_rilevati):
    lista_colori = []
    robot = np.array([0, 0])

    for i in range(len(punti_rilevati)):
        punto = punti_rilevati[i]
        
        # Calcolo distanza (Norma del vettore differenza)
        distanza = np.sqrt(np.sum((punto - robot)**2))
        
        # STRUTTURA CONDIZIONATA
        if distanza < 3.0:
            colore = '#ff0000' # Rosso
        elif distanza < 6.0:
            colore = '#ffcc00' # Giallo/Arancio
        else:
            colore = '#00ff00' # Verde
            
        lista_colori.append(colore)
        
    return lista_colori

# ==========================================================
# 3. AVVIO (Con correzione del Warning)
# ==========================================================
radar = RadarDinamico(n_oggetti=10)
radar.setup_radar()

# cache_frame_data=False elimina il warning del terminale
ani = FuncAnimation(radar.fig, radar.aggiorna, interval=100, 
                    blit=True, cache_frame_data=False)

plt.title("📡 RADAR TRACKING: Calcolo Distanze Euclidee", color='green', loc='left')
plt.show()
```
