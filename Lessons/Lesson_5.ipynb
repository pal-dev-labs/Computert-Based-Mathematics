     Le tarantole stanno ai criceti come le galline stanno ai gattini.
    

Un insieme di vettori che fanno tutti parte dello stesso insieme di dati è spesso chiamato *spazio vettoriale*

Lo spazio vettoriale degli animali in questa sezione ha due *dimensioni*, con cui si intende che ogni vettore nello spazio ha due numeri ad esso associati (cioè due colonne nel foglio di calcolo).

Il fatto che questo spazio abbia due dimensioni rende facile *visualizzare* lo spazio disegnando una trama 2D.

La maggior parte degli spazi vettoriali con cui si lavora ha più di due dimensioni, a volte molte centinaia.

In quei casi, è più difficile visualizzare lo "spazio", ma la matematica funziona più o meno allo stesso modo.

# Trovare il colore più vicino

Proprio come volevamo trovare l'animale che più si avvicinava a un punto arbitrario nello spazio carineria/dimensione, vorremmo trovare il nome del colore più vicino a un punto arbitrario nello spazio RGB.

Il modo più semplice per trovare l'elemento più vicino a un vettore arbitrario è semplicemente trovare la distanza tra il vettore di destinazione e ogni elemento nello spazio, a turno, quindi ordinare l'elenco dal più vicino al più lontano.

La funzione `closest()` qui sotto fa proprio questo. Per impostazione predefinita, restituisce un elenco dei dieci elementi più vicini al vettore specificato.

> Nota: il calcolo dei "vicini più vicini" in questo modo va bene per gli esempi in questo quaderno, ma ingestibile lento per spazi vettoriali di qualsiasi dimensione apprezzabile.

Man mano che il tuo spazio vettoriale cresce, vorrai passare a una soluzione più veloce, come [kdtree](https://docs.scipy.org/doc/scipy-0.14.0/reference/generated/scipy.spatial.KDTree) di SciPy .html) o [Annoy](https://pypi.python.org/pypi/annoy).

```python
points = dataframe_color.iloc[:,2:5].values
points
```

```python
# funzione che ci restituisce gli n vettori più vicini ad un vettore di riferimento dato ref_vector
def closest_indices(ref_vector, vectors,n):
    distances = np.sqrt(np.sum((vectors - ref_vector)**2, axis=1))
    # ordiniamo gli indici dei punti in base alla distanza dal vettore di riferimento e prendiamo i primi 10
    closest_indices = np.argsort(distances)[:n]
    return closest_indices
```

```python
# ESTRAGGO LA MATRICE DI COLORI
color_matrix=dataframe_color.iloc[:,2:5].values

# DEFIINISCO IL COLORE DI RIFERIMENTO
ref_vector = colors['red']

# DEFINISCO NUMERO DI VICINI
k_nearest= 10

cind = closest_indices(ref_vector,color_matrix,k_nearest)

# stampiamo i punti più vicini
print(cind)
```

```python

# 2. Pulizia manuale: eliminiamo dal testo i caratteri di punteggiatura
testo_pulito = testo_grezzo.lower()
for carattere in ".,:;!?":
    testo_pulito = testo_pulito.replace(carattere, "")

# 3. TOKENIZZAZIONE: Trasformazione in lista di parole
parole = testo_pulito.split()

# 4. Conteggio con un DICTIONARY
frequenza = {}

for p in parole:
    if p in frequenza:
        frequenza[p] += 1  # Se la parola esiste già, aumenta il contatore
    else:
        frequenza[p] = 1   # Se è la prima volta che la vediamo, impostala a 1

# 5. Visualizzazione dei risultati
print("Frequenza delle parole:")
for parola, conteggio in frequenza.items():
    print(f"{parola}: {conteggio}")
```

# PROVIAMO A VISUALIZZARE LA FREQUENZA DELLE SINGOLE PAROLE (PROBABILITA')

La Probabilità: È un concetto matematico fondamentale.

Se la parola "Dracula" appare 3 volte su un totale di 10 parole, ha una probabilità del $3/10 = 0.3$, ovvero il 30%.

$$P(parola) = \frac{\text{Conteggio parola}}{\text{Totale parole nell'opera}}$$

```python
import matplotlib.pyplot as plt

# 1. Calcolo il totale delle parole per la probabilità
totale_parole = sum(frequenza.values())

# 2. Ordino e prendo solo le prime 10
# sorted trasforma il dizionario in una lista di tuple (parola, conteggio) ordinata per valore
top_10 = sorted(frequenza.items(), key=lambda x: x[1], reverse=True)[:10]

# 3. Separo i dati per il grafico
parole_top = [item[0] for item in top_10]
conteggi_top = [item[1] for item in top_10]

# Calcolo la probabilità (frequenza relativa) in percentuale
# Formula: (conteggio_parola / totale_parole) * 100
probabilita_top = [(c / totale_parole) * 100 for c in conteggi_top]

# 4. Creazione del grafico
plt.figure(figsize=(10, 6))
barre = plt.bar(parole_top, conteggi_top, color='salmon', edgecolor='darkred')

# 5. Aggiungo la probabilità sopra ogni barra
for i, barra in enumerate(barre):
    altezza = barra.get_height()
    plt.text(barra.get_x() + barra.get_width()/2, altezza,
             f'{probabilita_top[i]:.1f}%',
             ha='center', va='bottom', fontweight='bold')

plt.title("Top 10 Parole più frequenti", fontsize=14)
plt.ylabel("Numero di occorrenze")
plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.show()
```

# "Le parole che contano"
In questa Top 10 ci sono molti articoli e congiunzioni.

Per rendere l'analisi davvero "Dantesca", è necessario aggiungere un altro filtro nel codice:

if not token.is_punct and not token.is_space and not token.is_stop

is_stop dice a spaCy di ignorare le

## "Stopwords" (parole comuni come il, lo, la, di, a, da).

Rilanciando il grafico con questo filtro, possiamo vedere parole come "occhi", "mondo", "stelle" o "gentile".

```python
import matplotlib.pyplot as plt
from collections import Counter

# 1. Estraiamo solo le PAROLE SIGNIFICATIVE
# Filtriamo:
# - no punteggiatura (.is_punct)
# - no spazi (.is_space)
# - no STOPWORDS (.is_stop): questo rimuove automaticamente articoli, congiunzioni, preposizioni
# - solo parole alfabetiche (.is_alpha): rimuove numeri o simboli strani
parole_significative = [
    token.text.lower()
    for token in doc
    if not token.is_punct
    and not token.is_space
    and not token.is_stop
    and token.is_alpha
    and len(token.text) > 2  # Escludiamo parole troppo corte (spesso rimasugli di scansione)
]

# 2. Conteggio delle frequenze sul nuovo elenco "pulito"
conteggio_frequenza = Counter(parole_significative)
totale_significative = len(parole_significative)

# 3. Prendiamo le prime 10 parole
top_10 = conteggio_frequenza.most_common(10)

# 4. Prepariamo i dati
nomi_parole = [item[0] for item in top_10]
valori_frequenza = [item[1] for item in top_10]
probabilita = [(v / totale_significative) * 100 for v in valori_frequenza]

# 5. Visualizzazione
plt.figure(figsize=(12, 7))
# Cambiamo colore in verde per indicare che i dati sono stati "filtrati/puliti"
barre = plt.bar(nomi_parole, valori_frequenza, color='seagreen', edgecolor='black')

# Aggiungiamo la probabilità sopra ogni barra
for i, barra in enumerate(barre):
    altezza = barra.get_height()
    plt.text(barra.get_x() + barra.get_width()/2, altezza + 2,
             f'{probabilita[i]:.2f}%', ha='center', va='bottom',
             fontweight='bold', color='darkgreen')

plt.title("Top 10 parole 'Chiave' nei promessi sposi (Senza Articoli/Congiunzioni)", fontsize=16)
plt.ylabel("Numero di occorrenze")
plt.xlabel("Parole Significative")
plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.show()
```

# Word Cloud (Nuvola di Parole)
È il modo più immediato per visualizzare i temi dominanti. Più una parola è grande, più è frequente. È perfetta per un impatto visivo immediato sui nomi propri o sui sostantivi.

```python
import spacy
from wordcloud import WordCloud
import matplotlib.pyplot as plt
from collections import Counter

# 1. Caricamento Modello
try:
    nlp = spacy.load("it_core_news_sm")
except:
    !python -m spacy download it_core_news_sm
    nlp = spacy.load("it_core_news_sm")

# 2. Lettura del file (assumendo che lo abbiamo scaricato prima)
with open("promessi_sposi.txt", "r", encoding="utf-8") as f:
    testo_completo = f.read()

# 3. Cerchiamo l'inizio reale per evitare metadati
inizio = testo_completo.find("Quel ramo del lago di Como")
# Analizziamo una fetta più corposa (i primi 3 capitoli, circa 150.000 caratteri)
testo_analisi = testo_completo[inizio:inizio+150000]

# 4. Elaborazione spaCy
doc = nlp(testo_analisi)

# 5. Estrazione Nomi Propri (Filtro doppio: NER e POS Tagging)
personaggi = []
for ent in doc.ents:
    if ent.label_ == "PER":
        nome = ent.text.strip().title()
        personaggi.append(nome)

# Se il NER fallisce, usiamo il POS Tagging come backup per i nomi propri
if not personaggi:
    personaggi = [token.text.title() for token in doc if token.pos_ == "PROPN" and len(token.text) > 2]

# 6. Pulizia e Filtro
stop_nomi = ["Don", "Donna", "Fra", "Signor", "Padre", "Sua", "Vostra", "Zio", "Zia", "Quel", "Cosa", "Santo", "Gutenberg"]
personaggi_filtrati = [p for p in personaggi if p not in stop_nomi]

# 7. Generazione WordCloud
if personaggi_filtrati:
    testo_stringa = " ".join(personaggi_filtrati)
    wordcloud = WordCloud(width=1000, height=500, background_color='white', colormap='plasma').generate(testo_stringa)

    plt.figure(figsize=(15, 8))
    plt.imshow(wordcloud, interpolation='bilinear')
    plt.axis("off")
    plt.title("Nuvola dei Personaggi (Manzoni)", fontsize=20)
    plt.show()
else:
    print("Errore: Ancora nessun nome trovato. Prova a stampare testo_analisi[:500] per verificare il contenuto.")
```

# Grafico a Torta (Distribuzione Grammaticale)
Questo serve per analizzare lo stile dell'autore. Manzoni usa più verbi (azione) o più aggettivi (descrizione)?

```python
# Contiamo le categorie grammaticali nel documento
pos_counts = Counter([token.pos_ for token in doc if not token.is_punct and not token.is_space])

# Selezioniamo solo le categorie principali per il grafico
labels = ['Nomi', 'Verbi', 'Aggettivi', 'Avverbi', 'Pronomi']
mapping = {'NOUN': 'Nomi', 'VERB': 'Verbi', 'ADJ': 'Aggettivi', 'ADV': 'Avverbi', 'PRON': 'Pronomi'}

values = [pos_counts.get(k, 0) for k in mapping.keys()]
display_labels = list(mapping.values())

plt.figure(figsize=(10, 8))
plt.pie(values, labels=display_labels, autopct='%1.1f%%', startangle=140, colors=plt.cm.Pastel1.colors)
plt.title("Lo stile di Manzoni: Composizione Grammaticale")
plt.show()
```

# Analisi linguistica e cromatica.

Dato che colors ha i nomi in inglese (es. 'blood red', 'sea green'), ma il romanzo di Manzoni è in italiano, il passaggio fondamentale è creare un ponte semantico. Non serve tradurre tutti i 500 colori: l'IA di spaCy ci permette di mappare i lemmi italiani alle categorie del tuo dizionario.

Cerchiamo di estrarre i colori e calcolare il vettore medio RGB del romanzo.

```python
import numpy as np
import spacy
import matplotlib.pyplot as plt
from collections import Counter

# 1. Carichiamo il modello italiano
nlp = spacy.load("it_core_news_sm")

# 2. Definiamo la mappatura Italiano -> Chiave del tuo dizionario 'colors'
# Abbiamo selezionato le tonalità più frequenti in Manzoni
mappa_manzoniana = {
    "nero": "black", "bianco": "white", "rosso": "red", "verde": "green",
    "azzurro": "sky blue", "blu": "blue", "giallo": "yellow", "marrone": "brown",
    "grigio": "grey", "fosco": "dark", "oscuro": "dark", "celeste": "light blue",
    "oro": "gold", "argenteo": "silver", "pallido": "pale", "rossastro": "reddish"
}

# 3. Analisi del testo (limitata per evitare errori di memoria)
doc = nlp(testo_romanzo[:100000])

vettori_colore = []
colori_rilevati = []

for token in doc:
    lemma = token.lemma_.lower()
    if lemma in mappa_manzoniana:
        nome_inglese = mappa_manzoniana[lemma]
        # Estraiamo il vettore array dal tuo dizionario 'colors'
        vettore_rgb = colors[nome_inglese]
        vettori_colore.append(vettore_rgb)
        colori_rilevati.append(lemma)
```

```python
# 4. Calcolo del Colore Medio tramite i vettori
if vettori_colore:
    # Calcoliamo la media lungo l'asse delle componenti (R, G, B)
    vettore_medio = np.mean(vettori_colore, axis=0).astype(int)

    # 5. Visualizzazione del Risultato
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(15, 6))

    # Grafico a barre delle frequenze
    conteggio = Counter(colori_rilevati)
    nomi_barre = list(conteggio.keys())
    valori_barre = list(conteggio.values())
    ax1.bar(nomi_barre, valori_barre, color='gray', alpha=0.7)
    ax1.set_title("Frequenza dei termini cromatici")
    ax1.tick_params(axis='x', rotation=45)

    # Box con il colore medio calcolato
    ax2.imshow([[vettore_medio]])
    ax2.set_title(f"Colore Medio del Romanzo\nRGB: {vettore_medio}")
    ax2.axis('off')

    plt.tight_layout()
    plt.show()
else:
    print("Nessun colore trovato con la mappatura attuale.")
```
