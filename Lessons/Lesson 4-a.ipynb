# CONCETTO: STRUTTURE CONDIZIONALI
Un programma è un elenco di istruzioni che vengono eseguite in sequenza.

Con le strutture condizionali, è possibile eseguire determinati blocchi di codice **CONDIZIONALMENTE** e/o **RIPETUTAMENTE**: questi blocchi di base possono essere combinati per creare programmi molto sofisticati.

## Riepilogo: Operazioni di confronto

Un tipo di operazione che può essere molto utile è il confronto tra valori diversi.
A tale scopo, Python implementa operatori di confronto standard, che restituiscono valori booleani ``True`` e ``False``.

Le operazioni di confronto sono elencate nella tabella seguente:

| ``a == b`` | ``a`` uguale a ``b``

| ``a != b`` | ``a`` diverso da ``b``

| ``a < b`` | ``a`` minore di ``b``

| ``a > b`` | ``a`` maggiore di ``b``

| ``a <= b`` | ``a`` minore o uguale a ``b``

|``a >= b`` | ``a`` maggiore o uguale a ``b``



## Conditional Statements: ``if``-``elif``-``else``:

Scriviamo una funzione Python che prende in input due liste e restituisce True se hanno almeno un elemento in comune.

```python
def checklist(lista1,lista2):
    found = False
    counter=0
    for i in lista1:
        for j in lista2:
            if i == j:
                counter=counter+1
                found = True
    return found,counter
```

```python
x = [1,2,3,5,6]
y= [1,2,7,8,9,10]
a1, a2 = checklist(x,y)
print("Trovato stesso elemento:", a1, " N. volte =", a2)
```

# LA DISTANZA EUCLIDEA

1. Consideriamo due punto nel piano $A=(x1,y1)$ e $B=(x2,y2)$
2. La distanza euclidea tra i due punti $d(A,B)$ è data da $$d = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$$
3. Un punto del piano $A(x,y)$ può essere associato ad un vettore $\vec v$ che ha come componenti le coordinate del punto $\vec v(x,y)$
4. I punti $A=(x1,y1)$ e $B=(x2,y2)$ possono quindi essere associati ai vettori $\vec v1(x_1,y_1), \vec v2(x_2,y_2)$
5. La differenza $\vec v1 - \vec v2$ è data da $[(x_2 - x_1), (y_2 - y_1)]$

6. Osserviamo che la differenza tra i vettori è proprio la differenza tra le coordinate dei punti A e B
7. Pertanto possiamo esprimere la distanza tra A e B come $d = |(\vec{v1}-\vec{v2})| = \sqrt{(x_2 - x_1)^2 + (y_2 - y_1)^2}$

I

```python
# punto_x1
x1 = np.array([1,2])

# punto_x2
x2 = np.array([5,7])

diff = x1-x2

print(diff)
```

```python
diff2 = diff**2
print(diff2)
```

```python
# 2. Calcolo della distanza Euclidea

# Eleviamo al quadrato le differenze, le sommiamo e facciamo la radice
distanza = np.sqrt(np.sum(diff2))
print(distanza)
```

# Alternativa: np.linalg.norm

```python
distanza2 = np.linalg.norm(x1 - x2)
print(distanza2)
```

# ESERCIZIO 2: moto nel piano

- Il corpo A si muove di moto parabolico. parte dal punto O(0m ,170m) con una velocita iniziale $\vec v_0=(12 m/s,7m/s)$.
$$\vec x(t) = (12 t, y(t) = 170 + 7 t - \frac{1}{2} g t^2)$$
- Il corpo B si muove con la seguente legge oraria: $$\vec x(t) = (30 cos(1 t), y(t) = 25 sin(1 t))$$

## Deermina la distanza minima tra i due corpi nei prima 10 secondi


```python
from scipy.optimize import minimize_scalar

def distanza(t):
    # Coordinate Corpo A
    xa = 12 * t
    ya = 170 + 7 * t - 0.5 * 9.81 * t**2

    # Coordinate Corpo B
    xb = 30 * np.cos(1 * t)
    yb = 25 * np.sin(1 * t)

    # Ritorna la distanza euclidea
    return np.sqrt((xa - xb)**2 + (ya - yb)**2)

# Cerchiamo il minimo nell'intervallo [0, 10] secondi
risultato = minimize_scalar(distanza, bounds=(0, 10), method='bounded')

print(f"Distanza minima: {risultato.fun:.4f} metri")
print(f"Raggiunta al tempo t: {risultato.x:.4f} secondi")
```

```python
from matplotlib.animation import FuncAnimation
from scipy.optimize import minimize_scalar
from IPython.display import HTML

# 1. Definizione dei parametri e delle funzioni orarie
t_max = 10
frames = 300
t_vettore = np.linspace(0, t_max, frames)
g = 9.81

# Funzione per calcolare la distanza a ogni istante t (necessaria per minimize_scalar)
def calcola_distanza(t):
    xa = 12 * t
    ya = 170 + 7 * t - 0.5 * g * t**2
    xb = 30 * np.cos(1 * t)
    yb = 25 * np.sin(1 * t)
    return np.sqrt((xa - xb)**2 + (ya - yb)**2)

# --- CALCOLO PRECISO CON MINIMIZE_SCALAR ---
risultato_ottimizzazione = minimize_scalar(calcola_distanza, bounds=(0, t_max), method='bounded')
t_min_esatto = risultato_ottimizzazione.x
d_min_esatta = risultato_ottimizzazione.fun

print(f"Distanza minima esatta: {d_min_esatta:.4f} m al tempo t = {t_min_esatto:.4f} s")

# Coordinate per il plot (vettorializzate)
xA = 12 * t_vettore
yA = 170 + 7 * t_vettore - 0.5 * g * t_vettore**2
xB = 30 * np.cos(1 * t_vettore)
yB = 25 * np.sin(1 * t_vettore)
distanze = calcola_distanza(t_vettore)

# 2. Setup della visualizzazione
fig, ax = plt.subplots(figsize=(10, 6))
ax.set_xlim(-40, 125)
ax.set_ylim(min(yA) - 20, max(yA) + 20)
ax.set_xlabel('X (m)')
ax.set_ylabel('Y (m)')
ax.set_title('Confronto Traiettorie e Minima Distanza')
ax.grid(True, linestyle='--', alpha=0.6)

lineA, = ax.plot([], [], 'r-', alpha=0.3, label='Scia A (Parabolico)')
pointA, = ax.plot([], [], 'ro', label='Corpo A')
lineB, = ax.plot([], [], 'b-', alpha=0.3, label='Scia B (Lissajous)')
pointB, = ax.plot([], [], 'bo', label='Corpo B')
segmento_dist, = ax.plot([], [], 'g--', label='Distanza Istantanea')

# Testi informativi
time_text = ax.text(0.05, 0.92, '', transform=ax.transAxes, weight='bold')
# Testo fisso per il minimo globale calcolato
ax.text(0.05, 0.82, f'Minimo Globale (SciPy):\n{d_min_esatta:.3f} m a {t_min_esatto:.3f} s',
        transform=ax.transAxes, color='darkgreen', bbox=dict(facecolor='white', alpha=0.5))

ax.legend(loc='upper right')

def init():
    lineA.set_data([], [])
    pointA.set_data([], [])
    lineB.set_data([], [])
    pointB.set_data([], [])
    segmento_dist.set_data([], [])
    time_text.set_text('')
    return pointA, pointB, lineA, lineB, segmento_dist, time_text

def update(i):
    pointA.set_data([xA[i]], [yA[i]])
    pointB.set_data([xB[i]], [yB[i]])
    lineA.set_data(xA[:i], yA[:i])
    lineB.set_data(xB[:i], yB[:i])
    segmento_dist.set_data([xA[i], xB[i]], [yA[i], yB[i]])

    time_text.set_text(f'Tempo: {t_vettore[i]:.2f} s\nDistanza corr: {distanze[i]:.2f} m')
    return pointA, pointB, lineA, lineB, segmento_dist, time_text

ani = FuncAnimation(fig, update, frames=frames, init_func=init, blit=True, interval=30)

# Rendering del video nel notebook
HTML(ani.to_html5_video())
```

```python
# 1. ABILITAZIONE WIDGET PER COLAB (Necessario eseguirlo all'inizio)
from google.colab import output
output.enable_custom_widget_manager()

# Comando magico per l'interattività
%matplotlib inline
# Nota: Su Colab 'widget' può essere instabile, spesso 'inline' con
# l'interattività dei widget di ipywidgets è la scelta più sicura.
# Ma useremo un approccio basato su ipywidgets per la massima compatibilità.

import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import minimize_scalar
import ipywidgets as widgets
from IPython.display import display

# --- LOGICA FISICA ---
g = 9.81
t_max = 10

def get_pos(t):
    xa = 12 * t
    ya = 170 + 7 * t - 0.5 * g * t**2
    xb = 30 * np.cos(1 * t)
    yb = 25 * np.sin(1 * t)
    return (xa, ya), (xb, yb)

def calcola_distanza(t):
    posA, posB = get_pos(t)
    return np.sqrt((posA[0] - posB[0])**2 + (posA[1] - posB[1])**2)

# --- CALCOLO MINIMO ---
res = minimize_scalar(calcola_distanza, bounds=(0, t_max), method='bounded')
t_min, d_min = res.x, res.fun

# --- SETUP GRAFICO ---
def plot_interattivo(t):
    fig, ax = plt.subplots(figsize=(10, 6))

    # Pre-calcolo traiettorie (scie)
    t_all = np.linspace(0, t_max, 500)
    posA_all, posB_all = get_pos(t_all)
    ax.plot(posA_all[0], posA_all[1], 'r-', alpha=0.15, label='Traiettoria A')
    ax.plot(posB_all[0], posB_all[1], 'b-', alpha=0.15, label='Traiettoria B')

    # Posizioni correnti al tempo t
    (xa, ya), (xb, yb) = get_pos(t)
    d = calcola_distanza(t)

    # Punti e linea distanza
    ax.plot(xa, ya, 'ro', markersize=10, label='Corpo A')
    ax.plot(xb, yb, 'bo', markersize=10, label='Corpo B')
    ax.plot([xa, xb], [ya, yb], 'g--', lw=2, label='Distanza')

    # Decorazioni
    ax.axhline(0, color='black', lw=1.5) # Terreno
    ax.set_xlim(-40, 130)
    ax.set_ylim(-200, 220) # Range ottimizzato per i tuoi dati
    ax.grid(True, alpha=0.3)
    ax.legend(loc='upper right')

    # Box informazioni
    info = f'Tempo: {t:.2f} s\nDistanza: {d:.2f} m'
    color_box = 'white'
    if abs(t - t_min) < 0.1:
        info += f'\n¡PUNTO DI MINIMO!'
        color_box = '#d4edda' # Verde chiaro

    ax.text(0.05, 0.95, info, transform=ax.transAxes, verticalalignment='top',
            bbox=dict(facecolor=color_box, alpha=0.8, edgecolor='gray'))

    plt.title("Simulazione Interattiva: Moto Corpi A e B")
    plt.show()

# --- CREAZIONE SLIDER (ipywidgets) ---
# Usiamo ipywidgets perché su Colab è molto più fluido di matplotlib.widgets
slider_t = widgets.FloatSlider(
    value=0,
    min=0,
    max=t_max,
    step=0.05,
    description='Tempo (s):',
    layout=widgets.Layout(width='80%')
)

# Visualizzazione
print(f"Punto di massima vicinanza calcolato: {d_min:.3f} m a t = {t_min:.3f} s")
widgets.interact(plot_interattivo, t=slider_t);
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML

# ==========================================================
# 1. MOTORE RADAR (Scatola nera)
# ==========================================================
class RadarSimulato:
    def __init__(self, n_oggetti=15):
        self.n = n_oggetti
        self.robot = np.array([0, 0])
        self.oggetti = np.random.uniform(-10, 10, (self.n, 2))
        self.velocita = np.random.uniform(-0.4, 0.4, (self.n, 2))

        # Setup Grafica
        self.fig, self.ax = plt.subplots(figsize=(8, 8))
        self.punti_grafici = self.ax.scatter([], [], s=100, edgecolors='black', zorder=10)

    def setup_estetica(self):
        """Configura il look del radar"""
        self.ax.set_xlim(-11, 11)
        self.ax.set_ylim(-11, 11)
        self.ax.set_facecolor('#f8f9fa')

        # Robot al centro
        self.ax.scatter(0, 0, color='blue', s=250, marker='s', label='Robot (Base)', zorder=15)

        # Zone di sicurezza
        for r, etichetta, col in zip([3, 6, 9], ['Pericolo', 'Attenzione', 'Sicuro'], ['red', 'orange', 'green']):
            cerchio = plt.Circle((0, 0), r, color=col, fill=False, linestyle='--', alpha=0.3, lw=1.5)
            self.ax.add_patch(cerchio)
            self.ax.text(0, r + 0.2, etichetta, color=col, fontsize=9, ha='center', weight='bold')

        self.ax.legend(loc='upper right')
        self.ax.grid(True, linestyle=':', alpha=0.5)
        self.ax.set_title("📡 SISTEMA RADAR ATTIVO - Monitoraggio Prossimità")

    def aggiorna_frame(self, frame):
        """Funzione di update per l'animazione"""
        # Muove gli oggetti
        self.oggetti += self.velocita

        # Rimbalzo sui bordi
        fuori_limite = (self.oggetti > 10.5) | (self.oggetti < -10.5)
        self.velocita[fuori_limite] *= -1

        # Logica degli studenti
        colori_aggiornati = analizza_prossimita(self.oggetti)

        # Applica i cambiamenti
        self.punti_grafici.set_offsets(self.oggetti)
        self.punti_grafici.set_color(colori_aggiornati)

        return self.punti_grafici,

# ==========================================================
# 2. AREA DI LAVORO STUDENTE
# ==========================================================
def analizza_prossimita(punti_rilevati):
    lista_colori = []
    posizione_robot = np.array([0, 0])

    for i in range(len(punti_rilevati)):
        punto_attuale = punti_rilevati[i]

        # Calcolo distanza euclidea
        distanza = np.sqrt(np.sum((punto_attuale - posizione_robot)**2))

        if distanza < 3.0:
            colore = 'red'
        elif distanza < 6.0:
            colore = 'orange'
        else:
            colore = 'green'

        lista_colori.append(colore)

    return lista_colori

# ==========================================================
# 3. ESECUZIONE (Ottimizzata per Colab)
# ==========================================================

# 1. Creiamo il radar
radar = RadarSimulato(n_oggetti=18)
radar.setup_estetica()

# 2. Creiamo l'animazione
# NOTA: Usiamo 'frames=100' per evitare che il rendering in Colab richieda troppo tempo
animazione = FuncAnimation(
    radar.fig,
    radar.aggiorna_frame,
    frames=100,
    interval=150,
    blit=True,
    cache_frame_data=False
)

# 3. IMPORTANTE PER COLAB: Nascondiamo il grafico statico e mostriamo il video
plt.close() # Chiude la figura 'fantasma' statica
HTML(animazione.to_html5_video()) # Genera il video interattivo
```
