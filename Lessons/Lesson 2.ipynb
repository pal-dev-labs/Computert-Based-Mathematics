```python
import matplotlib.pyplot as plt
import pandas as pd

def plot_players_by_nationality(df, top_n=20):
    """
    Crea un istogramma (bar chart) del numero di giocatori per nazione.
    
    Parameters:
    df (pd.DataFrame): Il dataframe dei giocatori.
    top_n (int): Il numero di nazioni da visualizzare (default 20).
    """
    # 1. Calcolo delle frequenze e ordinamento
    # value_counts() restituisce i dati già ordinati in modo decrescente
    nation_counts = df['nationality'].value_counts().head(top_n)
    
    # 2. Configurazione del grafico
    plt.figure(figsize=(14, 7))
    
    # Creazione delle barre
    nation_counts.plot(kind='bar', color='steelblue', edgecolor='black')
    
    # 3. Personalizzazione estetica
    plt.title(f'Top {top_n} Nazioni per Numero di Giocatori', fontsize=16, fontweight='bold')
    plt.xlabel('Nazione', fontsize=12)
    plt.ylabel('Conteggio Giocatori', fontsize=12)
    
    # Rotazione delle etichette per evitare sovrapposizioni
    plt.xticks(rotation=45, ha='right')
    
    # Aggiunta di una griglia leggera per la leggibilità sull'asse y
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    
    # Ottimizzazione del layout
    plt.tight_layout()
    
    # 4. Salvataggio dell'immagine
    filename = f'players_by_nation_top{top_n}.png'
    plt.savefig(filename)
    print(f"Grafico salvato con successo come: {filename}")

# Esempio di utilizzo:
plot_players_by_nationality(players, top_n=25)
```

# PROMPT GEMINI:

Scrivi una funzione che ha come input players per plottare con un pie chart il numero di giocatori per nazione

```python
import matplotlib.pyplot as plt

def plot_nationality_distribution(df, top_n=10):
    """
    Crea un pie chart della distribuzione dei giocatori per nazione.
    Raggruppa le nazioni oltre la top_n in una categoria 'Others'.
    
    Parameters:
    df (pd.DataFrame): Il dataframe dei giocatori.
    top_n (int): Quante singole nazioni mostrare prima di raggruppare (default 10).
    """
    # 1. Calcolo delle frequenze
    counts = df['nationality'].value_counts()
    
    # 2. Separazione: Top N vs Altri
    top_counts = counts.head(top_n)
    others_count = counts.iloc[top_n:].sum()
    
    # 3. Creazione dei dati finali per il grafico
    # Aggiungiamo la categoria 'Others' solo se ci sono effettivamente altre nazioni
    if others_count > 0:
        plot_data = top_counts.copy()
        plot_data['Others'] = others_count
    else:
        plot_data = top_counts

    # 4. Configurazione del grafico
    plt.figure(figsize=(8, 8))
    
    # Esplosione della prima fetta (la nazione con più giocatori) per evidenziarla
    explode = [0.1] + [0] * (len(plot_data) - 1)
    
    # Creazione del pie chart
    plot_data.plot(
        kind='pie', 
        autopct='%1.1f%%',   # Mostra la percentuale con un decimale
        startangle=140,      # Ruota il grafico per una migliore estetica
        explode=explode,     # Distacca la fetta principale
        shadow=True,         # Aggiunge un'ombra per profondità
        colors=plt.cm.Paired.colors, # Usa una palette di colori armoniosa
        label=''             # Rimuove l'etichetta di default dell'asse y
    )
    
    # 5. Titolo e Legenda
    plt.title(f'Distribuzione Giocatori per Nazione (Top {top_n} + Others)', fontsize=14, fontweight='bold')
    plt.axis('equal')  # Assicura che la torta sia un cerchio perfetto
    
    # Mostra la legenda a lato per non coprire il grafico
    plt.legend(plot_data.index, title="Nazioni", loc="center left", bbox_to_anchor=(1, 0, 0.5, 1))
    
    plt.tight_layout()
    plt.savefig('distribution_pie_chart.png')
    # plt.show()

# Esempio di chiamata:
plot_nationality_distribution(players, top_n=8)
```

## Per visualizzare come si distribuisce l'età all'interno dell'intero dataset, utilizzeremo un istogramma. Questo grafico è lo strumento statistico ideale perché raggruppa le età in "bin" (barre) 

## PROMPT GEMINI:  

Crea una funzione per plottare con un istogramma la distribuzione delle età per tutti i calciatori. evidenzia media varianza e deviazione stanrdar

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

def plot_age_distribution_with_stats(df):
    """
    Crea un grafico della distribuzione delle età indicando media, 
    varianza e il numero di giocatori nel range standard.
    """
    # 1. Calcoli Statistici
    mean_age = df['age'].mean()
    var_age = df['age'].var(ddof=0)
    std_age = df['age'].std(ddof=0)
    
    # Definiamo l'intervallo 'entro la varianza' (range di 1 deviazione standard)
    lower_bound = mean_age - std_age
    upper_bound = mean_age + std_age
    
    # Conteggio dei giocatori in questo range
    players_in_range = df[(df['age'] >= lower_bound) & (df['age'] <= upper_bound)].shape[0]
    
    # 2. Plotting
    plt.figure(figsize=(12, 7))
    
    # Istogramma con linea di densità
    sns.histplot(df['age'], bins=range(int(df['age'].min()), int(df['age'].max()) + 2), 
                 kde=True, color='firebrick', edgecolor='black', alpha=0.5)
    
    # Linea della Media
    plt.axvline(mean_age, color='blue', linestyle='--', linewidth=2, 
                label=f'Media: {mean_age:.2f}')
    
    # Area ombreggiata per indicare la dispersione (entro 1 Deviazione Standard)
    plt.axvspan(lower_bound, upper_bound, color='gray', alpha=0.2, 
                label=f'Range $\pm 1 \sigma$ (Varianza: {var_age:.2f})')
    
    # 3. Annotazione testuale con i dati richiesti
    stats_text = (f'Varianza ($\sigma^2$): {var_age:.2f}\n'
                  f'Dev. Std ($\sigma$): {std_age:.2f}\n'
                  f'Giocatori nel range: {players_in_range}')
    
    plt.text(0.95, 0.95, stats_text, transform=plt.gca().transAxes, 
             fontsize=11, verticalalignment='top', horizontalalignment='right',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    # Personalizzazione estetica
    plt.title('Distribuzione Età con Analisi della Varianza', fontsize=15, fontweight='bold')
    plt.xlabel('Età (Anni)', fontsize=12)
    plt.ylabel('Numero di Calciatori', fontsize=12)
    plt.legend(loc='upper left')
    plt.grid(axis='y', alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('distribuzione_eta_varianza.png')

# Esecuzione della funzione
plot_age_distribution_with_stats(players)
```

La varianza ($\sigma^2$ o $s^2$) è una misura statistica che indica quanto i valori di un insieme di dati sono dispersi o "sparpagliati" attorno alla loro media aritmetica.  

In termini semplici: 
- Se la varianza è bassa, i dati sono tutti molto vicini alla media (ad esempio, una squadra composta quasi interamente da ventiduenni).
- Se la varianza è alta, i dati sono molto distanti tra loro (ad esempio, una squadra con molti diciottenni e molti trentacinquenni).

La Formula della Varianza.  

La formula varia leggermente a seconda che si consideri l'intera popolazione (tutti i calciatori del database) o un campione (una selezione di essi).

$$\sigma^2 = \frac{\sum_{i=1}^{N} (x_i - \mu)^2}{N}$$

La Deviazione Standard (indicata con la lettera greca $\sigma$ o con la sigla S.D.) è la misura di dispersione più utilizzata in statistica. 

Se la varianza ti dice quanto i dati sono "sparpagliati" in termini di quadrati, la deviazione standard riporta quel valore all'unità di misura originale (nel tuo caso, gli anni).

1. La Formula MatematicaLa deviazione standard è semplicemente la radice quadrata della varianza:$$\sigma = \sqrt{\frac{\sum_{i=1}^{N} (x_i - \mu)^2}{N}}$$

2. La Regola Empirica (Distribuzione Normale. ): In una distribuzione "normale" (a campana), la deviazione standard ti permette di prevedere dove si trova la maggior parte dei tuoi giocatori: Circa il 68% dei giocatori avrà un'età compresa tra $(\mu - \sigma)$ e $(\mu + \sigma)$. Circa il 95% dei giocatori si troverà entro due deviazioni standard $(\mu \pm 2\sigma)$.

Esempio Pratico per la tua A.C. Roma 

Supponiamo che tu abbia acquistato i 6 giocatori e la loro età media sia 21 anni. 
- Scenario A (Bassa Deviazione Standard, $\sigma = 1$):I tuoi giocatori hanno quasi tutti la stessa età (es. tra i 20 e i 22 anni). La squadra è molto omogenea, cresceranno insieme, ma mancano "estremi" (né giovanissimi di 17 anni, né "esperti" di 23).
- Scenario B (Alta Deviazione Standard, $\sigma = 3$):Hai un gruppo molto variegato. La media è sempre 21, ma potresti avere un prodigio di 17 anni e un paio di ventitreenni già pronti. C'è più diversità di esperienza nel gruppo.

```python
import matplotlib.pyplot as plt
import seaborn as sns

def plot_height_distribution_with_stats(df):
    """
    Crea un grafico della distribuzione delle altezze indicando media, 
    deviazione standard e il numero di giocatori nel range.
    """
    # 1. Pulizia e Calcoli Statistici
    heights = df['height_cm'].dropna()
    
    mean_height = heights.mean()
    std_height = heights.std(ddof=0)
    
    # Intervallo entro 1 deviazione standard
    lower_bound = mean_height - std_height
    upper_bound = mean_height + std_height
    
    # Conteggio giocatori nel range
    players_in_range = df[(df['height_cm'] >= lower_bound) & (df['height_cm'] <= upper_bound)].shape[0]
    
    # 2. Creazione del Grafico
    plt.figure(figsize=(12, 7))
    
    # Istogramma con linea di densità (KDE)
    sns.histplot(heights, bins=range(int(heights.min()), int(heights.max()) + 2,1), 
                 kde=True, color='mediumpurple', edgecolor='black', alpha=0.6)
    
    # Linea della Media
    plt.axvline(mean_height, color='darkblue', linestyle='--', linewidth=2, 
                label=f'Media: {mean_height:.1f} cm')
    
    # Area della Deviazione Standard (Range di normalità)
    plt.axvspan(lower_bound, upper_bound, color='gray', alpha=0.2, 
                label=f'Range $\pm 1 \sigma$ (Dispersione)')
    
    # 3. Box Informativo
    stats_text = (f'Altezza Media: {mean_height:.1f} cm\n'
                  f'Dev. Std ($\sigma$): {std_height:.2f} cm\n'
                  f'Giocatori nel range: {players_in_range}')
    
    plt.text(0.95, 0.95, stats_text, transform=plt.gca().transAxes, 
             fontsize=11, verticalalignment='top', horizontalalignment='right',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    # Personalizzazione
    plt.title('Distribuzione dell\'Altezza dei Calciatori (cm)', fontsize=15, fontweight='bold')
    plt.xlabel('Altezza (cm)', fontsize=12)
    plt.ylabel('Frequenza (N. Giocatori)', fontsize=12)
    plt.legend(loc='upper left')
    plt.grid(axis='y', alpha=0.3)
    
    # Impostiamo i tick ogni 5 cm per una lettura rapida
    plt.xticks(range(int(heights.min()), int(heights.max()) + 1, 5))
    
    plt.tight_layout()
    plt.savefig('distribuzione_altezza_stats.png')
    print(f"Grafico salvato come distribuzione_altezza_stats.png")

# Esecuzione
plot_height_distribution_with_stats(players)
```

## MODELLO TEORICO: Curva di Gauss o Distribuzione Normale

Funzione di densità di probabilità della Normale: $$f(x) = \frac{1}{\sigma\sqrt{2\pi}} e^{-\frac{1}{2}\left(\frac{x-\mu}{\sigma}\right)^2}$$

$\mu$ (Media): Determina la posizione del picco (il "centro" dei calciatori).

$\sigma$ (Deviazione Standard): Determina quanto la campana è larga o stretta (la "precisione" della natura).

## PROMPT GEMINI

Crea una funzione che plotta la distribuzione normale e metta in evidenza la media, sigma, 2 sigma, 3 sigma

```python
import numpy as np
import scipy.stats as stats

def plot_gauss_comparison(df):
    plt.figure(figsize=(10, 6))
    
    # 1. Dati reali (Istogramma)
    data = df['height_cm'].dropna()
    mu, std = data.mean(), data.std()
    
    # Plot istogramma normalizzato (density=True per confrontarlo con la curva)
    plt.hist(data, bins=30, density=True, alpha=0.6, color='skyblue', edgecolor='black')

    # 2. Curva di Gauss teorica
    xmin, xmax = plt.xlim()
    x = np.linspace(xmin, xmax, 100)
    p = stats.norm.pdf(x, mu, std) # La funzione matematica di Gauss
    
    plt.plot(x, p, 'r', linewidth=3, label='Curva di Gauss Teorica')
    
    plt.title(f"Fit Matematico: $\mu={mu:.1f}$, $\sigma={std:.2f}$")
    plt.legend()
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

def plot_normal_distribution_rules(mu=181, std=7):
    """
    Plotta la curva di Gauss evidenziando le zone di confidenza (1, 2, 3 sigma).
    Parametri di default basati sull'altezza media dei calciatori (181cm, std 7).
    """
    # 1. Definizione del range dell'asse x (copriamo 4 sigma per vedere bene le code)
    x = np.linspace(mu - 4*std, mu + 4*std, 1000)
    y = stats.norm.pdf(x, mu, std)

    plt.figure(figsize=(14, 8))
    plt.plot(x, y, color='black', linewidth=2, label='Curva di Gauss')

    # 2. Colorazione delle aree di Sigma
    # Area entro 3 sigma (99.7%)
    plt.fill_between(x, y, where=((x >= mu - 3*std) & (x <= mu + 3*std)), 
                     color='lightgray', alpha=0.3, label='3 Sigma (99.7%)')
    
    # Area entro 2 sigma (95%)
    plt.fill_between(x, y, where=((x >= mu - 2*std) & (x <= mu + 2*std)), 
                     color='darkgray', alpha=0.4, label='2 Sigma (95.4%)')
    
    # Area entro 1 sigma (68%)
    plt.fill_between(x, y, where=((x >= mu - std) & (x <= mu + std)), 
                     color='skyblue', alpha=0.5, label='1 Sigma (68.3%)')

    # 3. Linee verticali per i marker
    plt.axvline(mu, color='blue', linestyle='--', label=f'Media ($\mu$): {mu}')
    
    # Aggiunta etichette per i vari sigma sull'asse x
    sigmas = [mu - 3*std, mu - 2*std, mu - std, mu + std, mu + 2*std, mu + 3*std]
    labels = ['-3σ', '-2σ', '-1σ', '+1σ', '+2σ', '+3σ']
    plt.xticks(list(plt.xticks()[0]) + sigmas, rotation=45)

    # 4. Personalizzazione
    plt.title('Distribuzione Normale: Regola del 68-95-99.7', fontsize=16, fontweight='bold')
    plt.xlabel('Valore (es. Altezza in cm)', fontsize=12)
    plt.ylabel('Densità di Probabilità', fontsize=12)
    plt.legend()
    plt.grid(axis='y', alpha=0.2)
    
    plt.tight_layout()
    plt.show()

# Esecuzione della funzione
plot_normal_distribution_rules()
```

## PROMPT GEMINI

Crea una funzione la distribuzione dell'altezza dei calciatori e la sovrapponga alla curva teorica della distribuzione di gauss 

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import scipy.stats as stats

def plot_height_with_gauss(df):
    """
    Plotta l'istogramma reale delle altezze e vi sovrappone 
    la curva teorica di Gauss calcolata sui parametri del dataset.
    """
    # 1. Preparazione dati (pulizia dai valori nulli)
    data = df['height_cm'].dropna()
    
    # 2. Calcolo dei parametri della popolazione
    mu = data.mean()       # Media reale
    sigma = data.std()     # Deviazione standard reale
    
    # 3. Configurazione del grafico
    plt.figure(figsize=(12, 7))
    
    # Istogramma REALE normalizzato (densità)
    # L'area totale delle barre sarà pari a 1 per permettere il confronto con Gauss
    sns.histplot(data, stat="density", color="skyblue", alpha=0.5, 
                 label="Dati Reali (Frequenza relativa)", edgecolor="white")
    
    # 4. Creazione della curva di GAUSS teorica
    # Creiamo un range di valori x che copre l'altezza minima e massima
    x = np.linspace(data.min(), data.max(), 200)
    # Calcoliamo la PDF (Probability Density Function) per ogni x
    y = stats.norm.pdf(x, mu, sigma)
    
    # Sovrapposizione della curva teorica
    plt.plot(x, y, color='firebrick', linewidth=3, label='Curva di Gauss Teorica')

    # 5. Evidenziatore della Media
    plt.axvline(mu, color='navy', linestyle='--', alpha=0.8, 
                label=f'Media ($\mu$): {mu:.1f} cm')

    # 6. Personalizzazione e Titoli
    plt.title('Altezza Calciatori: Confronto Dati Reali vs Modello di Gauss', fontsize=15, fontweight='bold')
    plt.xlabel('Altezza (cm)', fontsize=12)
    plt.ylabel('Densità di Probabilità', fontsize=12)
    plt.legend()
    plt.grid(axis='y', alpha=0.2)
    
    plt.tight_layout()
    plt.savefig('gauss_vs_reale_altezza.png')
    # plt.show()

# Esempio di utilizzo:
# plot_height_with_gauss(players)
```

CREA una funzione Plotta ora la distribuzione dei salari

```python
import matplotlib.pyplot as plt
import seaborn as sns

def plot_wage_distribution_with_stats(df):
    """
    Crea un grafico della distribuzione dei salari indicando media, 
    deviazione standard e il numero di giocatori nel range.
    """
    # 1. Pulizia e Calcoli Statistici
    # Rimuoviamo i valori nulli per evitare errori nel calcolo
    wages = df['wage_eur'].dropna()
    
    mean_wage = wages.mean()
    std_wage = wages.std(ddof=0)
    
    # Intervallo entro 1 deviazione standard
    lower_bound = max(0, mean_wage - std_wage) # Il salario non può essere negativo
    upper_bound = mean_wage + std_wage
    
    # Conteggio giocatori nel range
    players_in_range = df[(df['wage_eur'] >= lower_bound) & (df['wage_eur'] <= upper_bound)].shape[0]
    
    # 2. Plotting
    plt.figure(figsize=(12, 7))
    
    # Per una migliore visualizzazione, limitiamo l'asse X al 99° percentile 
    # (evita che i pochi top player schiaccino tutto il grafico a sinistra)
    limit = wages.quantile(0.99)
    sns.histplot(wages[wages <= limit], bins=50, kde=True, color='gold', edgecolor='black', alpha=0.6)
    
    # Linea della Media
    plt.axvline(mean_wage, color='red', linestyle='--', linewidth=2, 
                label=f'Media: €{mean_wage:,.0f}')
    
    # Area della Deviazione Standard
    plt.axvspan(lower_bound, min(limit, upper_bound), color='gray', alpha=0.2, 
                label=f'Range $\pm 1 \sigma$')
    
    # 3. Annotazione Statistica
    stats_text = (f'Media: €{mean_wage:,.2f}\n'
                  f'Dev. Std ($\sigma$): €{std_wage:,.2f}\n'
                  f'Giocatori nel range: {players_in_range}')
    
    plt.text(0.95, 0.85, stats_text, transform=plt.gca().transAxes, 
             fontsize=11, verticalalignment='top', horizontalalignment='right',
             bbox=dict(boxstyle='round', facecolor='white', alpha=0.8))

    # Personalizzazione
    plt.title('Distribuzione dei Salari Settimanali (€)', fontsize=15, fontweight='bold')
    plt.xlabel('Salario (EUR)', fontsize=12)
    plt.ylabel('Frequenza (N. Giocatori)', fontsize=12)
    plt.legend(loc='upper right')
    plt.grid(axis='y', alpha=0.3)
    
    plt.tight_layout()
    plt.savefig('distribuzione_salari_stats.png')
    print(f"Grafico salvato come distribuzione_salari_stats.png")

# Esecuzione
plot_wage_distribution_with_stats(players)
```

GEMINI SCRIVI una funzione per plottare VALORE DEL CARTELLINO VS OVERALL

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

def plot_value_vs_overall(df):
    """
    Crea uno scatter plot tra Overall e Valore di mercato,
    aggiungendo una linea di tendenza per visualizzare la crescita esponenziale.
    """
    # 1. Pulizia dati: filtriamo solo chi ha un valore di mercato (escludiamo svincolati)
    df_clean = df[df['value_eur'] > 0].dropna(subset=['overall', 'value_eur'])
    
    plt.figure(figsize=(12, 8))
    
    # 2. Scatter Plot
    # Usiamo un alpha basso (0.4) perché ci sono migliaia di punti sovrapposti
    sns.scatterplot(data=df_clean, x='overall', y='value_eur', 
                    alpha=0.4, color='teal', edgecolor=None)
    
    # 3. Linea di tendenza (Regressione polinomiale di 2° grado)
    # Una retta non basterebbe a descrivere questo mercato, serve una curva
    z = np.polyfit(df_clean['overall'], df_clean['value_eur'], 2)
    p = np.poly1d(z)
    
    x_range = np.linspace(df_clean['overall'].min(), df_clean['overall'].max(), 100)
    plt.plot(x_range, p(x_range), color='red', linewidth=3, label='Trend di Mercato')
    
    # 4. Personalizzazione assi
    plt.title('Relazione tra Valore di Mercato e Overall', fontsize=16, fontweight='bold')
    plt.xlabel('Overall (Abilità Attuale)', fontsize=12)
    plt.ylabel('Valore del Cartellino (€)', fontsize=12)
    
    # Formattiamo l'asse Y per leggere meglio i milioni
    plt.ticklabel_format(style='plain', axis='y')
    
    plt.grid(True, linestyle='--', alpha=0.3)
    plt.legend()
    
    plt.tight_layout()
    #plt.savefig('value_vs_overall.png')
    plt.show()

# Esecuzione
plot_value_vs_overall(players)
```

GEMINI SCRIVI una funzione per plottare WAGE_EUR SU BASE ANNUALE VS OVERALL

```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

def plot_annual_wage_vs_overall(df):
    """
    Crea un grafico che mette in relazione il salario annuale (calcolato)
    con l'overall dei calciatori.
    """
    # 1. Pulizia dati e calcolo del salario annuale
    # Filtriamo i valori nulli e i salari pari a zero
    df_clean = df[(df['wage_eur'] > 0) & (df['overall'] > 0)].copy()
    
    # Calcolo: Salario Settimanale * 52 settimane
    df_clean['annual_wage_eur'] = df_clean['wage_eur'] * 52
    
    # 2. Configurazione del grafico
    plt.figure(figsize=(12, 8))
    
    # Scatter plot con colore sfumato in base all'overall
    scatter = plt.scatter(df_clean['overall'], df_clean['annual_wage_eur'], 
                          alpha=0.5, c=df_clean['overall'], cmap='viridis', s=30)
    
    # 3. Aggiunta della linea di tendenza (Regressione di 2° grado)
    # Serve a mostrare come il costo annuo esploda per i giocatori elite
    z = np.polyfit(df_clean['overall'], df_clean['annual_wage_eur'], 2)
    p = np.poly1d(z)
    x_range = np.linspace(df_clean['overall'].min(), df_clean['overall'].max(), 100)
    plt.plot(x_range, p(x_range), color='red', linewidth=3, label='Trend Salario Annuale')

    # 4. Personalizzazione estetica e assi
    plt.title('Salario Annuale Stimato vs Overall', fontsize=16, fontweight='bold')
    plt.xlabel('Overall (Abilità Attuale)', fontsize=12)
    plt.ylabel('Salario Annuale (€)', fontsize=12)
    
    # Formattazione asse Y per mostrare i milioni in modo leggibile
    plt.ticklabel_format(style='plain', axis='y')
    
    plt.colorbar(scatter, label='Livello Overall')
    plt.grid(True, linestyle='--', alpha=0.3)
    plt.legend()
    
    plt.tight_layout()
    plt.savefig('annual_wage_vs_overall.png')
    # plt.show()

# Esempio di utilizzo:
plot_annual_wage_vs_overall(players)
```

Check out the complete list of all FIFA positions now:

Goalkeeper (GOAL / GK)

The Goalkeeper, represented by the acronym GOL (or GK in English – Goalkeeper), is one of the most important positions in the game.

Whether in career mode or playing online matches in Ultimate Team, it is essential that you always have two goalkeepers at your disposal: one on the field and the other on the bench.

That’s because we know that anything can happen within FIFA matches. So, in case your goalkeeper gets injured or ends up getting a red card, it’s important that you have another one on the bench to replace him.

Centre-Back (ZAG ​​/ CB)

Defenders, in turn, represented by the acronym ZAG (CB in English – Central Back), form the most important part of your defense on the field.

It’s interesting that you have another one on the bench in case something happens to one of those on the field, especially if you play with a lineup with two defenders.

An alternative is to have two full-backs on the bench, as many players in that position can also act as full-backs without losing overall points.

Right Back (LD / RB)

The right side, represented by the acronym LD (RB in English – Right Back), is the one who takes care of the defense of the right side of the field.

The vast majority of players who play in this position can also be selected to play in the left-back position.

Some of these players can also be cast in games as defenders or as left midfielders or right midfielders, as they perform well both in attack and defense.

In FIFA 21, full-backs who can also play as a right midfielder or left midfielder were represented by the acronym ADD.

However, in FIFA 22 this acronym no longer exists. In any case, check the SoFIFA website. It is the best option to know which players can be selected in these positions.

Left Back (LE / LB)

The left side, represented by the acronym LE (LB in English – Left Back), is the player responsible for defending the left side of the field.

Like right-backs, left-backs can also be cast as right-backs if the need arises.

Some of these players, as well as the right backs, can also be called as defenders or as left or right midfielders.

In any case, just follow the tip we gave in the previous topic and stay tuned to SoFIFA.

Central Defensive Midfielder (VOL / CDM)

The defensive midfielder, represented by the acronym VOL (CDM in English – Central Defensive Midfielder) acts as a more defensive midfielder, but in the end, ends up acting as a handyman in the midfield.

Central Midfielder (MC/CM)

The midfielder, represented by the acronym MC (CM in English – Central Midfielder) is responsible for setting up the plays.

In very difficult cases, you can put a right midfielder, left midfielder or attacking midfielder to play in this position if necessary.

Right Midfielder (MD/RM)

The Right Midfielder, represented by the acronym MD (RM in English – Right Midfielder) is a slightly more versatile player, who can be scaled in place of the wingers of your team.

He can be replaced by the left midfielder without major problems. In some cases, he may lose some points, but there are players who do well in both positions.

He’s a more attacking midfielder type, but he doesn’t do as well in the more central positions of the field.

If necessary, he can be replaced by a winger or an attacking midfielder.

Left Midfielder (ME/LM)

The left midfielder, represented by the acronym ME (LM in English – Left Midfielder) is a player who is in the offensive part on the left.

Like the right midfielder, he can also play in the opposite position, and there are several players who are versatile and can play on both sides of the field.

Like the right midfielder, these players just don’t play very well in the most central offensive position (although there are exceptions, such as Jadon Sancho) and can be replaced by players who act as wingers.

Central Attacking Midfielder (MEI / CAM)

The Offensive Midfielder, represented by the acronym MEI (CAM in English – Central Attacking Midfielder), in turn, is a type of midfielder who plays more advanced, in a more suitable position for the attack.

They are the most versatile players in the game, as they can replace left midfielders, right midfielders, left wingers, right wingers and centre forwards.

Left Wing (PE / LW)

The left winger, represented by the acronym PE (LW in English – Left Wing) is an attacking player who plays on the left. This is the case of Neymar and Vinícius Jr, for example, and they are some of the fastest players in the game.
